# [Python Tutorial for Beginners 4: Lists, Tuples, and Sets](https://www.youtube.com/watch?v=W8KRzm-HUcc&list=PL-osiE80TeTt2d9bfVyTiXJA-UTHn6WwU&index=4)

```python
course = ['History', 'Math', 'Physics', 'CompSci']
course2 = ['Art', 'Education']
print(course) # ['History', 'Math', 'Physics', 'CompSci']

print(course[0]) # History
print(course[-1]) # CompSci
print(course[0:2]) # ['History', 'Math']

course.append('Art')
print(course) # ['History', 'Math', 'Physics', 'CompSci', 'Art

course.insert(0, 'Art')
course.remove('Art')
course.remove('Art')
course.insert(0, course2)
print(course) # [['Art', 'Education'], 'History', 'Math', 'Physics', 'CompSci']

course.pop() # removes last item
print(course) # [['Art', 'Education'], 'History', 'Math', 'Physics']    

nums = [1, 5, 2, 4, 3]
nums.sort()
print(nums) # [1, 2, 3, 4, 5]
nums.sort(reverse=True)
print(nums) # [5, 4, 3, 2, 1]

sorted(nums) # returns a new sorted list      
print(min(nums)) # 1
print(max(nums)) # 5
print(sum(nums)) # 15

print(course.index('Math')) # 2
print('Math' in course) # True
print('Art' in course) # False

for i, n in enumerate(course): #enumerate returns index and value of the list
    print(i, n) # 0 ['Art', 'Education'], 1 'History', 2 'Math', 3 'Physics'
    
for i, n in enumerate(course, start=1): #enumerate returns index and value of the list
    print(i, n) # 1 ['Art', 'Education'], 2 'History', 3 'Math', 4 'Physics'
    
course_str = ', '.join(map(str, course))  # converts all list items to strings
print(course_str)
print(course_str) # ['Art', 'Education'], History, Math, Physics
```

```python
course = {'Math', 'Physics', 'CompSci'}
course2 = {'Art', 'Education'}
print(course) # {'Math', 'Physics', 'CompSci'}
print(type(course)) # <class 'set'>
print(course.intersection(course2)) # set() - returns common items in both sets
print(course.difference(course2)) # {'Math', 'Physics', 'CompSci'} -
print(course.union(course2)) # {'Math', 'Physics', 'CompSci', 'Art', 'Education'} - returns all items in both sets
```
