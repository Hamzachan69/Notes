# [Python Tutorial for Beginners 3: Integers and Floats - Working with Numeric Data](https://www.youtube.com/watch?v=khKv-8q7YmY&list=PL-osiE80TeTt2d9bfVyTiXJA-UTHn6WwU&index=3)


```python
n = 3 
print(type(n)) 
n = 3.14
print(type(n)) 

print(10 + 3) # 13
print(10 - 3) # 7
print(10 * 3) # 30
print(10 / 3) # 3.3333333333333335
print(10 // 3) # 3
print(10 % 3) # 1
print(10 ** 3) # 1000
print(10 + 3 * 2) # 16
print((10 + 3) * 2) # 26


num = 1 
num += 1 # num = num + 1
print(num) # 2
num -= 1 # num = num - 1
print(num) # 1
num *= 2 # num = num * 2
print(num) # 2

print(abs(-5)) # 5
print(round(3.7)) # 4
print(round(3.212344, 1)) # 3.2

n1 = 2 
n2 = 3

print(n1 > n2) # False
print(n1 < n2) # True
print(n1 >= n2) # False
print(n1 <= n2) # True
print(n1 == n2) # False
print(n1 != n2) # True
n1 = '2' 
n2 = '3'

print(n1 + n2)
n1 = int('2') 
n2 = int('3')
print(n1 + n2)
```
